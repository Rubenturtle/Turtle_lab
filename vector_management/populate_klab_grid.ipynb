```python
import geopandas as gpd
```

```python

# Load your grid (replace with your file path, e.g. a shapefile or GeoPackage)
grid = gpd.read_file(r"C:\Users\ruben.crespo\Downloads\test_grid.shp")

# Parameters
top_left_index = 62282
cols = 358
rows = 128
step_down = 360  # index decreases when moving south

# Get centroid coordinates of each tile
grid["centroid"] = grid.geometry.centroid
grid["lon"] = grid["centroid"].x
grid["lat"] = grid["centroid"].y

# Find reference min/max to compute row and col
min_lon = grid["lon"].min()
max_lat = grid["lat"].max()

# Compute row and col for each polygon
grid["row"] = ((max_lat - grid["lat"]).round()).astype(int)   # 0 = top
grid["col"] = ((grid["lon"] - min_lon).round()).astype(int)   # 0 = left

# Compute tile index
def compute_tile_index(row, col, top_left=top_left_index):
    return top_left - row * step_down + col

grid["tile_index"] = grid.apply(lambda x: compute_tile_index(x["row"], x["col"]), axis=1)

# Save updated file
grid.drop(columns=["centroid"], inplace=True)

# Save updated file
grid.to_file(r"C:\Users\ruben.crespo\Downloads\test_grid_1.shp")
```
