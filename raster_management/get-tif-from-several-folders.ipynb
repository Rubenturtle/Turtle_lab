```python
import os
import shutil
```

```python
def copy_rasters(source_dir, dest_dir, raster_extensions=(".tif", ".tiff", ".img")):
    """
    Iterates over folders in 'source_dir', finds raster files,
    and copies them to 'dest_dir'.

    Args:
        source_dir (str): Path to the root directory containing folders.
        dest_dir (str): Path where copied rasters will be stored.
        raster_extensions (tuple): Extensions considered as rasters.
    """

    # Create destination folder if it doesn't exist
    os.makedirs(dest_dir, exist_ok=True)

    # Walk through all subdirectories
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            # Check if file extension matches raster types
            if file.lower().endswith(raster_extensions):
                src_path = os.path.join(root, file)
                dest_path = os.path.join(dest_dir, file)

                # Avoid overwriting by adding suffix if needed
                if os.path.exists(dest_path):
                    base, ext = os.path.splitext(file)
                    count = 1
                    while os.path.exists(dest_path):
                        dest_path = os.path.join(dest_dir, f"{base}_{count}{ext}")
                        count += 1

                # Copy the raster
                shutil.copy2(src_path, dest_path)
                print(f"Copied: {src_path} → {dest_path}")

    print("\n✅ Done! All raster files have been copied.")
```

```python
source_directory = r"Z:\z_resources\caterina\valencia\hazard\mitigrate_flood_depth\merged"
# Don't set the folder in the same location as the source.
destination_directory = r"Z:\z_resources\caterina\valencia\hazard\mitigrate_flood_depth\merged\output_rasters"
copy_rasters(source_directory, destination_directory)
```
