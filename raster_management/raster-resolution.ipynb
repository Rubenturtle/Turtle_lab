```python
from osgeo import gdal
```

```python
input_raster = r"Z:\z_resources\un_gbf\01_aggregation_phase\02_mask_rasters\global_wetlands_20m_2020.tif"
output_raster = r"Z:\z_resources\un_gbf\01_aggregation_phase\02_mask_rasters\global_wetlands_100m_2020.tif"
```

```python
ds = gdal.Open(input_raster)
gt = ds.GetGeoTransform()

# Current resolution
x_res = gt[1]
y_res = abs(gt[5])

# New resolution
gdal.Warp(
    output_raster,
    input_raster,
    xRes=0.0009,  # ~100m at equator approx
    yRes=0.0009,
    resampleAlg="average",
    creationOptions=["COMPRESS=DEFLATE", "TILED=YES"]
)

print("Done")
```
